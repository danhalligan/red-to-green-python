## The data

This is a small dataset compiled from the results of a Facebook poll. Each row is one respondent, who was asked how they like three things:

| Column | Question | Response scale |
|---|---|---|
| `Steak` | How do you like your steak? | 1 (rarest) to 5 (most well done) |
| `Toast` | How do you like your toast? | 1 (lightest) to 6 (most burnt) |
| `Tea` | How do you like your tea? | `a` (darkest) to `f` (milkiest) |

The scales are shown in the poll image below.

![Steak, toast, and tea poll response scales](../assets/steak-toast-tea-poll.png)
