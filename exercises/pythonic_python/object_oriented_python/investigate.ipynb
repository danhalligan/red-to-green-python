## Meet the band

Ten animals, each with a name that belongs to the instance and a sound that
belongs to the class. `str()` shows the name; `lyric()` never uses it. Is the
output what you expected for every animal?

```python
band = [
    Dog("Elvis Pawsley"),
    Cat("Meowly Cyrus"),
    Bird("Chick Jagger"),
    Mouse("Rodent Stewart"),
    Cow("Moodonna"),
    Frog("Snoop Froggy Frog"),
    Elephant("Trunk Sinatra"),
    Duck("Beakyoncé"),
    Fish("Mackerelmore"),
    Seal("Notorious F.L.I.P."),
]

for animal in band:
    print(f"{animal!s:<32}{lyric(animal)}")
```

## The robot cover band

A `Robot` is not an `Animal`, but `lyric` and `sing` don't mind. `cover`

sing every line. Predict the output, including the last four lines.

```python
marvin = Robot("Marvin the Paranoid Android")
print("\n".join(cover(band, marvin)))
```
